# 200-Qubit Monolithic Planar Star-Fluxonium Processor

Here, a single planar die holds a 10 × 20 array of high-kinetic-inductance integer-fluxonium qubits. Each star-shaped capacitor shunts an aluminium Josephson junction in parallel with a granular-aluminium superinductor. The qubit spectra target zero externally applied flux, $\Phi_{\mathrm{ext}}/\Phi_0=0$, at the integer-fluxonium operating point [1, 2].

Fixed single-junction transmons mediate all 370 nearest-neighbor edges, with 190 horizontal `TransmonPocket` devices and 180 vertical `TransmonPocket6` devices. A capacitive microwave port drives each coupler through a `CoupledLineTee`, following the microwave-activation scheme analyzed in [7]. The architecture has zero dedicated flux-bias lines and zero superconducting quantum interference device (SQUID) tuners. On-die airbridges provide signal crossovers and ground straps.

## First-Principles Primer

Electric and magnetic fields around two coplanar waveguide (CPW) center traces overlap most strongly where the traces run parallel, with mutual capacitance and inductance changing the even and odd propagation modes [9]. In the 33 × 27.5 mm passive-link mask, a section through the closest vertical pair gives 33.36 µm of ground between the CPW slots, 44.36 µm between signal edges, and approximately 100 µm of parallel run. For ideal, uniform 8.5 µm traces with 5.5 µm slots on silicon under air, a quasi-static cross-section calculation gives about 2.4 fF/mm of mutual capacitance at 33.36 µm of ground and 0.14 fF/mm at 200 µm. These values describe the electric contribution per length for parallel sections, while the field distribution, terminations, and package set the actual microwave coupling.

Wen introduced the coplanar waveguide as a transmission line whose signal strip and ground share a plane [10]. The name transmon combines transmission line and plasma oscillation, following the capacitively shunted circuit described by Koch and colleagues [11]. Wang and colleagues analyze how a microwave-driven transmon couples neighboring integer-fluxonium circuits [7].

The 40.5 × 34 mm layout uses twenty readout/control feedlines and thirty charge-drive through lines, with one hundred signal launchers and protected ground-bond landings. Each vertical edge uses two capacitive access leads and a fixed transmon between them, so the vertical graph holds all 180 nearest-neighbor connections. The north, south, and side ports fit the native pocket geometry, which permits the required three connections with the access lines outside the readout corridor.

Here, the phase operator $\hat{\varphi}$ and its conjugate charge operator $\hat n$ dictate the dynamics of the nominal lumped fluxonium system. Electrostatic costs of adding a single Cooper pair to the island are quantified by the charging energy $E_C$. In parallel, the Josephson energy $E_J$ determines the maximum dissipationless tunneling current across the junction. Inductive energy $E_L$ defines the magnetic energy stored in the high-kinetic-inductance shunt. According to quantum mechanics, the ratio of the Planck constant $h$ to the elementary charge $e$ generates the flux quantum $\Phi_0=h/(2e)$ serving as the fundamental scaling unit of magnetic flux. These parameters formulate the Hamiltonian describing the system restricted to the lowest-energy bands.

$$
\hat H=4E_C\hat n^2+\frac{E_L}{2}\left(\hat{\varphi}-2\pi\frac{\Phi_{\mathrm{ext}}}{\Phi_0}\right)^2-E_J\cos\hat{\varphi}.
$$

The kinetic sheet inductance $L_{\square}$ scales linearly with the geometric ratio of the wire length $\ell$ to the wire width $w$. This scaling yields the nominal shunt inductance $L_k$ in the limit of negligible geometric inductance. Alternatively, the junction current density $J_c$ dictates the required overlap area $A_J$. Measured process values set the corresponding nominal critical current $I_c$.

$$
L_k=L_{\square}\frac{\ell}{w}, \qquad E_L=\frac{(\Phi_0/2\pi)^2}{L_k}, \qquad C_{\Sigma}=\frac{e^2}{2E_C}, \qquad I_c=\frac{2\pi E_J}{\Phi_0}, \qquad A_J=\frac{I_c}{J_c}.
$$

Zero-flux calculation compares two phase grids to check numerical convergence. Capacitance extraction facilitates the derivation of the charging energy of the drawn star electrodes directly from the physical layout. A thin-film eigenmode solver identifies wire self-resonances and helps bound the frequency range of the lumped approximation [2, 3]. Material controls enforce nominal input parameters. Measured sheet inductance and junction current density dictate the process-specific wire lengths and overlap areas for physical fabrication.

Each native `RouteMeander` component integrates its circular bends directly into the total path. The tee contribution includes the circular-bend length correction so that the nominal electrical path follows the drawn centerline. The nominal readout model operates with a phase velocity $v_p=1.20\times10^8$ m/s bounded by a capacitive qubit-facing end and a shorted coupler termination. Frequency-separated microwave tones drive shared control operations within the common readout network. The resonator-mediated drive mechanism follows circuit quantum electrodynamics [4], with drive strengths set by the extracted circuit and calibrated microwave pulses.

$$
\ell_r=\frac{v_p}{4f_r}=\ell_{\mathrm{meander}}+\ell_{\mathrm{coupler}}, \qquad f_r\propto\ell_r^{-1}.
$$

## Microwave-Driven Coupling

A fixed transmon supplies a nonlinear intermediate mode, in which a charge drive addresses transitions whose frequencies depend on the neighboring fluxonium states. Conditional excitation and return of that mode provide a microwave route to an entangling phase [7], while one physical junction fixes each coupler's Josephson energy without magnetic-flux tuning. Two qubit ports face opposite islands, and the microwave port couples to the lower island.

Each charge-drive primary CPW connects through the tee primary pins, while the open-ended secondary CPW couples across the native gap and connects to a transmon charge port. The coupling section holds an 8 µm ground strip between the CPW slots, and native routing supplies circular bends to the secondary and its access branch. Extracted drive-network scattering parameters determine the delivered microwave amplitude and the loading of each transmon.

Nineteen coupler frequencies share each row-wise charge-drive line, and nine share each column-wise line. Junction-process scatter shifts those frequencies. Pulse calibration on the fabricated spectrum identifies usable frequency windows and spectator constraints. Ground spacing separates the charge-drive trunks from the readout returns, and the automatic row pitch provides room for the coupling pockets and their access traces.

This design follows the fixed-transmon microwave option analyzed for integer fluxonium by Wang and colleagues [7]. Heunisch and colleagues model a related architecture with tunable SQUID couplers [8]. Electrostatic extraction sets the effective capacitance matrix, after which dressed spectra, spectator interactions, drive selectivity, and pulse calibration determine the usable gate parameters.

$$
\hat H_{\mathrm{FTF}}/h=\sum_{i=1,c,2}\hat H_i/h+\sum_{i<j}J_{ij}\hat n_i\hat n_j, \qquad E_{C,i}/h=\frac{e^2(C^{-1})_{ii}}{2h}, \qquad J_{ij}=\frac{4e^2(C^{-1})_{ij}}{h}.
$$

Here $C$ denotes the effective mode-capacitance matrix after reducing the floating islands, and $J_{ij}$ denotes the coefficient of the charge-product interaction in frequency units.

## Vertical Pocket Sizing

Charging-energy target of $E_{C,c}/h=0.300$ GHz corresponds to $C_{\Sigma,c}=64.57$ fF for the differential mode. The planar electrostatic model sweeps island width with a 20 µm island gap and 90 µm island height, holding the 650 × 300 µm ground pocket fixed. It solves the charge-free slots at an air–silicon interface with zero ground potential, using 5.0 µm and 2.5 µm grids to check numerical sensitivity [12].

$$
C_{\mathrm{pad}}=\frac{e^2}{2E_{C,c}}-c_J A_J-C_{\mathrm{port}}.
$$

```python
# --- Zero-Bias Integer-Fluxonium Targets ---
phi0 = h / (2 * e)
ej_ghz, ec_ghz = 6.78, 1.47
el_classes_ghz = np.linspace(0.20, 0.29, 10)
capacitance_target_ff = e**2 / (2 * h * ec_ghz * 1e9) * 1e15
critical_current_na = 2 * np.pi * h * ej_ghz * 1e9 / phi0 * 1e9
junction_area_um2 = (
    critical_current_na * 1e-9 / (JUNCTION_CURRENT_DENSITY_A_CM2 * 1e4) * 1e12
)
inductances_nh = (phi0 / (2 * np.pi)) ** 2 / (h * el_classes_ghz * 1e9) * 1e9
assert SHEET_INDUCTANCE_PH > 0 and JUNCTION_CURRENT_DENSITY_A_CM2 > 0

# Finite-difference spectra use nominal lumped parameters at Phi_ext = 0.
spectrum_records = []
for label, el_ghz in enumerate(el_classes_ghz):
    spectra = []
    for count in (2401, 4801):
        phase = np.linspace(-8 * np.pi, 8 * np.pi, count)
        step = phase[1] - phase[0]
        diagonal = 8 * ec_ghz / step**2 + el_ghz * phase**2 / 2 - ej_ghz * np.cos(phase)
        off_diagonal = np.full(count - 1, -4 * ec_ghz / step**2)
        energies = eigh_tridiagonal(
            diagonal, off_diagonal, select="i", select_range=(0, 4), eigvals_only=True
        )
        spectra.append(energies - energies[0])
    assert np.max(np.abs(spectra[1] - spectra[0])) < 0.003
    spectrum_records.append(
        dict(
            frequency_class=label,
            EJ_GHz=ej_ghz,
            EC_GHz=ec_ghz,
            EL_GHz=el_ghz,
            L_nH=inductances_nh[label],
            C_target_fF=capacitance_target_ff,
            f01_GHz=spectra[1][1],
            f12_GHz=spectra[1][2] - spectra[1][1],
        )
    )
spectrum_table = pd.DataFrame(spectrum_records)
assert np.min(np.diff(spectrum_table.f01_GHz)) > 0.050
display(spectrum_table.round(5))
```

```python
# --- Fixed Single-Junction Transmon Targets ---
coupler_ec_ghz = 0.300
coupler_ej_classes = np.linspace(32.0, 46.0, 19)
coupler_capacitance_ff = e**2 / (2 * h * coupler_ec_ghz * 1e9) * 1e15
charge = np.arange(-35, 36, dtype=float)
coupler_records = []
for label, energy in enumerate(coupler_ej_classes):
    levels = eigh_tridiagonal(
        4 * coupler_ec_ghz * charge**2,
        np.full(len(charge) - 1, -energy / 2),
        select="i",
        select_range=(0, 5),
        eigvals_only=True,
    )
    current_na = 2 * np.pi * h * energy * 1e9 / phi0 * 1e9
    area_um2 = current_na * 1e-9 / (JUNCTION_CURRENT_DENSITY_A_CM2 * 1e4) * 1e12
    coupler_records.append(
        dict(
            coupler_class=label,
            EJ_GHz=energy,
            EC_GHz=coupler_ec_ghz,
            C_target_fF=coupler_capacitance_ff,
            f01_GHz=levels[1] - levels[0],
            anharmonicity_GHz=levels[2] - 2 * levels[1] + levels[0],
            junction_area_um2=area_um2,
        )
    )
coupler_spectrum = pd.DataFrame(coupler_records)
assert np.min(np.diff(coupler_spectrum.f01_GHz)) > 0.080

# --- Representative Fluxonium-Transmon-Fluxonium Spectra ---
# These capacitance targets require electrostatic extraction from the drawn masks.
contact_ff, stray_ff, island_ground_ff = 3.0, 0.010, 10.0
interisland_ff = coupler_capacitance_ff - island_ground_ff / 2
node_capacitance = np.array(
    [
        [capacitance_target_ff, -contact_ff, 0, -stray_ff],
        [
            -contact_ff,
            interisland_ff + island_ground_ff + contact_ff,
            -interisland_ff,
            0,
        ],
        [
            0,
            -interisland_ff,
            interisland_ff + island_ground_ff + contact_ff,
            -contact_ff,
        ],
        [-stray_ff, 0, -contact_ff, capacitance_target_ff],
    ]
)
# Eliminate the floating common mode at zero common-mode charge.
transform = np.array([[1, 0, 0, 0], [0, 0.5, 0, 1], [0, -0.5, 0, 1], [0, 0, 1, 0]])
mode_capacitance = transform.T @ node_capacitance @ transform
cap_matrix_ff = (
    mode_capacitance[:3, :3]
    - np.outer(mode_capacitance[:3, 3], mode_capacitance[3, :3])
    / mode_capacitance[3, 3]
)
assert np.min(np.linalg.eigvalsh(cap_matrix_ff)) > 0
charging_matrix = e**2 * np.linalg.inv(cap_matrix_ff * 1e-15) / (2 * h * 1e9)
triplet_records = []
for axis, left_class, right_class, coupler_class in (
    ("H", 0, 1, 0),
    ("H", 8, 9, 18),
    ("V", 3, 0, 0),
    ("V", 9, 6, 4),
):
    checks = []
    for nf, nc in ((16, 5), (20, 7)):
        modes, charges = [], []
        for mode, label in enumerate((left_class, coupler_class, right_class)):
            ec_eff = charging_matrix[mode, mode]
            if mode == 1:
                values, vectors = eigh_tridiagonal(
                    4 * ec_eff * charge**2,
                    np.full(len(charge) - 1, -coupler_ej_classes[label] / 2),
                    select="i",
                    select_range=(0, nc - 1),
                )
                number = vectors.T @ (charge[:, None] * vectors)
            else:
                phase = np.linspace(-8 * np.pi, 8 * np.pi, 4801)
                step = phase[1] - phase[0]
                values, vectors = eigh_tridiagonal(
                    8 * ec_eff / step**2
                    + el_classes_ghz[label] * phase**2 / 2
                    - ej_ghz * np.cos(phase),
                    np.full(len(phase) - 1, -4 * ec_eff / step**2),
                    select="i",
                    select_range=(0, nf - 1),
                )
                phase_matrix = vectors.T @ (phase[:, None] * vectors)
                number = (
                    1j
                    * (values[:, None] - values[None, :])
                    * phase_matrix
                    / (8 * ec_eff)
                )
            modes.append(values - values[0])
            charges.append(number)
        identities = [np.eye(nf), np.eye(nc), np.eye(nf)]
        hamiltonian = np.diag(
            (
                modes[0][:, None, None]
                + modes[1][None, :, None]
                + modes[2][None, None, :]
            ).ravel()
        ).astype(complex)
        for a, b in ((0, 1), (1, 2), (0, 2)):
            operators = identities.copy()
            operators[a], operators[b] = charges[a], charges[b]
            hamiltonian += (
                8
                * charging_matrix[a, b]
                * np.kron(np.kron(operators[0], operators[1]), operators[2])
            )
        energies, states = eigh(hamiltonian, check_finite=False)
        labels = [(i, c, j) for i in range(2) for c in range(2) for j in range(2)]
        indices = [(i * nc + c) * nf + j for i, c, j in labels]
        populations = abs(states[indices, :]) ** 2
        bare_rows, dressed_columns = linear_sum_assignment(-populations)
        assigned = dict(zip((labels[i] for i in bare_rows), dressed_columns))
        dressed = {label: energies[column] for label, column in assigned.items()}
        transitions = np.array(
            [
                dressed[(i, 1, j)] - dressed[(i, 0, j)]
                for i in range(2)
                for j in range(2)
            ]
        )
        zz_khz = (
            dressed[(1, 0, 1)]
            + dressed[(0, 0, 0)]
            - dressed[(1, 0, 0)]
            - dressed[(0, 0, 1)]
        ) * 1e6
        overlap = min(
            populations[labels.index((i, 0, j)), assigned[(i, 0, j)]]
            for i in range(2)
            for j in range(2)
        )
        checks.append((transitions, zz_khz, overlap))
    convergence_khz = max(
        np.max(abs(checks[1][0] - checks[0][0])) * 1e6,
        abs(checks[1][1] - checks[0][1]),
    )
    assert convergence_khz < 100.0, (left_class, convergence_khz)
    assert checks[1][2] > 0.90, (left_class, checks[1][2])
    triplet_records.append(
        dict(
            axis=axis,
            left_class=left_class,
            right_class=right_class,
            coupler_class=coupler_class,
            static_ZZ_kHz=checks[1][1],
            min_computational_overlap=checks[1][2],
            drive_00_GHz=checks[1][0][0],
            target_separation_MHz=np.min(abs(checks[1][0][1:] - checks[1][0][0])) * 1e3,
            min_branch_spacing_MHz=np.min(np.diff(np.sort(checks[1][0]))) * 1e3,
            truncation_change_kHz=convergence_khz,
        )
    )
triplet_table = pd.DataFrame(triplet_records)
display(coupler_spectrum.round(5))
display(triplet_table.round(5))
print(
    "Nominal lumped spectra; extracted capacitances and calibrated pulses set gate performance."
)
```

```python
# --- Vertical Pocket Capacitance ---
pocket_width_um, pocket_height_um = 650, 300
island_height_um, island_gap_um = 90, 20
junction_specific_cap_ff_um2 = 45.0
port_loading_ff = 1.5
silicon_relative_permittivity = 11.7
width_samples_um = np.array([330, 370, 410])
capacitance_samples = []
for mesh_um in (5.0, 2.5):
    grid_x, grid_y = int(1280 / mesh_um), int(640 / mesh_um)
    x = (np.arange(grid_x) + 0.5 - grid_x / 2) * mesh_um
    y = (np.arange(grid_y) + 0.5 - grid_y / 2) * mesh_um
    wave_number = np.hypot(fftfreq(grid_y)[:, None], rfftfreq(grid_x))
    pocket = (abs(x) < pocket_width_um / 2) & (
        abs(y[:, None]) < pocket_height_um / 2
    )
    grid_capacitances = []
    for width_um in width_samples_um:
        islands = (abs(x) < width_um / 2) & (
            (abs(y[:, None]) > island_gap_um / 2)
            & (abs(y[:, None]) < island_height_um + island_gap_um / 2)
        )
        potential = np.where(islands, np.sign(y[:, None]) / 2, 0.0)
        free = np.flatnonzero(pocket & ~islands)
        operator = LinearOperator(
            (len(free), len(free)),
            matvec=lambda voltage: irfft2(
                wave_number
                * rfft2(
                    np.bincount(free, weights=voltage, minlength=grid_x * grid_y)
                    .reshape(grid_y, grid_x)
                )
            ).ravel()[free],
            dtype=float,
        )
        source_charge = irfft2(wave_number * rfft2(potential)).ravel()[free]
        slot_voltage, status = cg(operator, -source_charge, rtol=1e-7)
        assert status == 0, "Pocket electrostatic solution did not converge."
        potential.ravel()[free] = slot_voltage
        charge = irfft2(wave_number * rfft2(potential))
        grid_capacitances.append(
            (1 + silicon_relative_permittivity)
            * epsilon_0
            * 2
            * np.pi
            * mesh_um
            * 1e9
            * np.sum(charge * potential)
        )
    capacitance_samples.append(grid_capacitances)

coarse_cap_ff, fine_cap_ff = np.array(capacitance_samples)
assert np.max(abs(coarse_cap_ff - fine_cap_ff)) < 2.0
assert np.all(np.diff(fine_cap_ff) > 0)
junction_cap_ff = (
    junction_specific_cap_ff_um2 * coupler_spectrum.junction_area_um2.to_numpy()
)
pad_target_ff = coupler_capacitance_ff - junction_cap_ff - port_loading_ff
assert np.all((fine_cap_ff[0] < pad_target_ff) & (pad_target_ff < fine_cap_ff[-1]))
vertical_width_um = (5 * np.rint(
    np.interp(pad_target_ff, fine_cap_ff, width_samples_um) / 5
)).astype(int)
vertical_cap_ff = (
    np.interp(vertical_width_um, width_samples_um, fine_cap_ff)
    + junction_cap_ff
    + port_loading_ff
)
assert np.max(abs(vertical_cap_ff - coupler_capacitance_ff)) < 0.5
display(pd.DataFrame({
    "class": coupler_spectrum.coupler_class,
    "island_width_um": vertical_width_um,
    "junction_capacitance_fF": junction_cap_ff,
    "estimated_total_capacitance_fF": vertical_cap_ff,
}).round(3))
```

```python
# --- Fixed Transmon Couplers on Both Lattice Axes ---
transmon_couplers, coupler_links, coupler_physical_records = [], [], []
horizontal_banks = [[] for _ in range(rows)]
vertical_banks = [[] for _ in range(columns)]
edge_specs = []
for row in range(rows):
    for col in range(columns - 1):
        edge_specs.append(
            (
                "H",
                row,
                col,
                qubits[row * columns + col],
                qubits[row * columns + col + 1],
            )
        )
for row in range(rows - 1):
    for col in range(columns):
        edge_specs.append(
            (
                "V",
                row,
                col,
                qubits[(row + 1) * columns + col],
                qubits[row * columns + col],
            )
        )
for axis, row, col, a, b in edge_specs:
    x = (float(a.p.pos_x) + float(b.p.pos_x)) / 2
    y = (
        (float(a.p.pos_y) + float(b.p.pos_y)) / 2 - 0.100
        if axis == "V"
        else float(a.p.pos_y) - 0.150
    )
    label = (col + 5 * row) % 19 if axis == "H" else (2 * row + 3 * col) % 19
    connection_pads = Dict()
    locations = (
        (("left", -1, 1), ("right", 1, -1), ("drive", -1, -1))
        if axis == "H"
        else (("left", 0, 1), ("right", 0, -1), ("drive", -1, -1))
    )
    for name, west, north in locations:
        connection_pads[name] = Dict(
            loc_W=str(west),
            loc_H=str(north),
            pad_width="60um",
            pad_height="15um",
            pad_gap="15um",
            pad_cpw_shift="5um",
            pad_cpw_extent="25um",
            pocket_rise="0um",
            pocket_extent="5um",
            cpw_extend="100um",
            cpw_width="8.5um",
            cpw_gap="5.5um",
        )
    component_type = TransmonPocket if axis == "H" else TransmonPocket6
    tc = component_type(
        design,
        f"TC_{axis}_{row}_{col}",
        options=Dict(
            pos_x=f"{x}mm",
            pos_y=f"{y}mm",
            pad_width=f"{vertical_width_um[label] if axis == 'V' else 455}um",
            pad_height=f"{island_height_um}um",
            pad_gap=f"{island_gap_um}um",
            pocket_width=f"{pocket_width_um}um",
            pocket_height=f"{pocket_height_um}um",
            inductor_width="6um",
            connection_pads=connection_pads,
        ),
    )
    transmon_couplers.append(tc)
    (horizontal_banks[row] if axis == "H" else vertical_banks[col]).append(tc)
    area_um2 = float(coupler_spectrum.iloc[label].junction_area_um2)
    overlap = area_um2 * 1e-6 / 0.0002
    assert 0 < overlap < 0.001
    junctions = design.qgeometry.tables["junction"]
    design.qgeometry.tables["junction"] = junctions[junctions.component != tc.id].copy()
    lower = draw.Polygon(
        [
            (-0.004, -0.023),
            (0.004, -0.023),
            (0.004, -0.008),
            (0.0001, -0.001),
            (0.0001, 0),
            (-0.0001, 0),
            (-0.0001, -0.001),
            (-0.004, -0.008),
        ]
    )
    upper = draw.Polygon(
        [
            (-0.0001, -overlap),
            (0.0001, -overlap),
            (0.0001, 0.001),
            (0.004, 0.008),
            (0.004, 0.023),
            (-0.004, 0.023),
            (-0.004, 0.008),
            (-0.0001, 0.001),
        ]
    )
    bandages = draw.union(
        *[draw.rectangle(0.006, 0.006, 0, yy) for yy in (-0.020, 0.020)]
    )
    marker = draw.rectangle(0.0002, overlap, 0, -overlap / 2)
    junction = draw.LineString([(0, -overlap), (0, 0)])
    lower, upper, bandages, marker, junction = draw.translate(
        [lower, upper, bandages, marker, junction], x, y
    )
    for name, geometry, layer in (
        ("junction_lower", lower, 3),
        ("junction_upper", upper, 4),
        ("bandage_contacts", bandages, 5),
        ("junction_overlap", marker, 6),
    ):
        tc.add_qgeometry("poly", {name: geometry}, layer=layer)
    tc.add_qgeometry("junction", {"JJ": junction}, width=0.0002, layer=3)
    coupler_physical_records.append(
        dict(
            coupler=tc.name,
            axis=axis,
            row=row,
            column=col,
            coupler_class=label,
            left_qubit=a.name,
            right_qubit=b.name,
            model_f01_GHz=float(coupler_spectrum.iloc[label].f01_GHz),
            nominal_junction_area_um2=area_um2,
            island_width_um=1000 * tc.p.pad_width,
            estimated_vertical_capacitance_fF=(
                vertical_cap_ff[label] if axis == "V" else np.nan
            ),
        )
    )
    for q, qp, cp in (
        ((a, "pin_cpl1", "left"), (b, "pin_cpl3", "right"))
        if axis == "H"
        else ((a, "pin_cpl4", "left"), (b, "pin_cpl2", "right"))
    ):
        pin_inputs = Dict(
            start_pin=Dict(component=q.name, pin=qp),
            end_pin=Dict(component=tc.name, pin=cp),
        )
        if axis == "V":
            link = RouteStraight(
                design,
                f"QCLINK_{axis}_{q.name}_{tc.name}",
                options=Dict(
                    trace_width="8.5um", trace_gap="5.5um", pin_inputs=pin_inputs
                ),
            )
        else:
            p0, p1 = q.pins[qp].middle, tc.pins[cp].middle
            jog_x = (
                float(q.p.pos_x) - cpw_pitch - cpw_gap_mm
                if p0[0] > p1[0] else (p0[0] + p1[0]) / 2
            )
            link = RouteAnchors(
                design,
                f"QCLINK_{axis}_{q.name}_{tc.name}",
                options=Dict(
                    trace_width="8.5um",
                    trace_gap="5.5um",
                    fillet="10um",
                    pin_inputs=pin_inputs,
                    anchors=Dict(
                        {0: np.array([jog_x, p0[1]]), 1: np.array([jog_x, p1[1]])}
                    ),
                    lead=Dict(start_straight="20um", end_straight="20um"),
                ),
            )
        coupler_links.append(link)
coupler_physical_table = pd.DataFrame(coupler_physical_records)
print(len(transmon_couplers), "fixed transmon couplers on both axes.", flush=True)

# --- Multiplexed Readout and Microwave Access ---
readout_frequencies = np.linspace(6.6, 8.0, 10)
phase_velocity = 1.20e8
for row in range(rows):
    for col in range(columns):
        index = row * columns + col
        q = qubits[index]
        x, y = float(q.p.pos_x), float(q.p.pos_y)
        tee = CoupledLineTee(
            design,
            f"CLT{index}",
            options=Dict(
                pos_x=f"{x + 0.425}mm",
                pos_y=f"{y + readout_height}mm",
                prime_width="8.5um",
                prime_gap="5.5um",
                second_width="8.5um",
                second_gap="5.5um",
                coupling_length="250um",
                coupling_space="8um",
                down_length="180um",
                fillet="50um",
                open_termination=False,
            ),
        )
        secondary = tee.qgeometry_table("path").query("name == 'second_cpw'").iloc[0]
        arm_length = secondary.geometry.length - (2 - np.pi / 2) * secondary.fillet
        target_length = phase_velocity / (4 * readout_frequencies[col % 10] * 1e9) * 1e3
        route = RouteMeander(
            design,
            f"RO_{index}",
            options=Dict(
                trace_width="8.5um",
                trace_gap="5.5um",
                fillet="50um",
                total_length=f"{target_length - arm_length}mm",
                pin_inputs=Dict(
                    start_pin=Dict(component=q.name, pin="pin_rdout"),
                    end_pin=Dict(component=tee.name, pin="second_end"),
                ),
                lead=Dict(
                    start_straight="300um",
                    end_straight="150um",
                    start_jogged_extension=Dict(
                        {0: ["R45", "300um"], 1: ["L90", "150um"]}
                    ),
                ),
                meander=Dict(spacing="150um", asymmetry="-100um"),
                prevent_short_edges="true",
            ),
        )
        couplers.append(tee)
        resonators.append(route)
        length_records.append(
            dict(
                qubit=q.name,
                target_GHz=readout_frequencies[col % 10],
                length_mm=route.length + arm_length,
            )
        )
print(len(resonators), "coupled readout resonators.", flush=True)
```

```python
# --- Orthogonal Feedlines and Edge Launchpads ---
feedline_connections, feedline_turns, feedline_routes = [], [], []
horizontal_drive_y = [
    bank[0].pins.drive.middle[1] - cpw_pitch - drive_tee_pitch - cpw_gap_mm
    for bank in horizontal_banks
]
for row in range(rows):
    y = float(qubits[row * columns].p.pos_y) + readout_height
    for half, side in enumerate((-1, 1)):
        indices = list(
            range(
                row * columns + half * (columns // 2),
                row * columns + (half + 1) * (columns // 2),
            )
        )
        if side == 1:
            indices.reverse()
        entry, exit = (
            ("prime_start", "prime_end") if side == -1 else ("prime_end", "prime_start")
        )
        tag = f"{row}_{half}"
        group = [couplers[i] for i in indices]
        cap = CapNInterdigital(
            design,
            f"CAP_{tag}",
            options=Dict(
                pos_x=f"{side * (chip_x / 2 - 1.1)}mm",
                pos_y=f"{y}mm",
                orientation=str(side * 90),
                north_width="8.5um",
                north_gap="5.5um",
                south_width="8.5um",
                south_gap="5.5um",
                cap_width="10um",
                cap_gap="6um",
                cap_gap_ground="6um",
                finger_length="20um",
                finger_count="5",
                cap_distance="50um",
            ),
        )
        for label, py in (("IN", y), ("OUT", y + port_pitch)):
            pad = LaunchpadWirebond(
                design,
                f"LP_FEED_{tag}_{label}",
                options=Dict(
                    pos_x=f"{side * (chip_x / 2 - .570)}mm",
                    pos_y=f"{py}mm",
                    orientation="0" if side == -1 else "180",
                    trace_width="8.5um",
                    trace_gap="5.5um",
                    lead_length="25um",
                    pad_width="170um",
                    pad_height="230um",
                    taper_height="240um",
                    pad_gap="70um",
                ),
            )
            launchpads.append(pad)
            name = f"FEED_{tag}_{label}"
            if label == "IN":
                a, ap, b, bp = pad, "tie", cap, "south_end"
                points = [a.pins[ap].middle, b.pins[bp].middle]
            else:
                a, ap, b, bp = group[-1], exit, pad, "tie"
                p0, p1 = a.pins[ap].middle, b.pins[bp].middle
                turn_x = p0[0] - side * 0.400
                return_y = (
                    min(py, horizontal_drive_y[row + 1] - cpw_pitch - cpw_gap_mm)
                    if row + 1 < rows else py
                )
                points = [p0, [turn_x, y], [turn_x, return_y]]
                if return_y < py:
                    lead_x = p1[0] - side * 0.100
                    points.extend([[lead_x - side * 0.150, return_y], [lead_x, py]])
                points.append(p1)
                feedline_turns.append(
                    dict(
                        name=name,
                        x_mm=turn_x,
                        y_bottom_mm=y,
                        y_top_mm=return_y,
                        vertical_leg=draw.LineString([[turn_x, y], [turn_x, return_y]]),
                        reference=draw.LineString(points),
                    )
                )
            feedline_connections.append((name, a, ap, b, bp, points))
        joints = [(cap, "north_end", group[0], entry)]
        joints += [(a, exit, b, entry) for a, b in zip(group[:-1], group[1:])]
        for index, (a, ap, b, bp) in enumerate(joints):
            feedline_connections.append(
                (
                    f"FEED_{tag}_{index}",
                    a,
                    ap,
                    b,
                    bp,
                    [a.pins[ap].middle, b.pins[bp].middle],
                )
            )

# Preserve the orthogonal corner coordinates when inserting native airbridges.
link_names = {link.id: link.name for link in coupler_links}
link_paths = design.qgeometry.tables["path"]
link_paths = link_paths[link_paths.component.isin(link_names) & ~link_paths["subtract"]]
link_tree = STRtree(link_paths.geometry.to_list())
for name, a, ap, b, bp, points in feedline_connections:
    hits = link_tree.query(draw.LineString(points), predicate="intersects")
    crossed_links = tuple(link_names[cid] for cid in link_paths.iloc[hits].component)
    items = chip.split_at_crossings(design, points, name, over=crossed_links, tag="ABF")
    chain, pieces = [], []
    for kind, geometry in items:
        if kind == "bridge":
            chain.append((design.components[geometry], "a", "b"))
        else:
            piece_name = name + (f"_{len(pieces)}" if pieces else "")
            route = chip.PolylineCPW(
                design,
                piece_name,
                options=Dict(
                    points=[list(map(float, point)) for point in geometry],
                    trace_width="8.5um",
                    trace_gap="5.5um",
                    fillet="35um",
                    min_segment="30um",
                ),
            )
            pieces.append(route)
            feedline_routes.append(route)
            chain.append((route, "start", "end"))
    for left, right in zip(chain[:-1], chain[1:]):
        design.connect_pins(left[0].id, left[2], right[0].id, right[1])
    design.connect_pins(a.id, ap, chain[0][0].id, chain[0][1])
    design.connect_pins(chain[-1][0].id, chain[-1][2], b.id, bp)
    if name.endswith("_OUT"):
        turn = next(item for item in feedline_turns if item["name"] == name)
        for route in pieces:
            path = route.qgeometry_table("path").query("subtract == False").iloc[0]
            assert turn["reference"].buffer(1e-6).covers(path.geometry), route.name
print("Readout launchers", len(launchpads), flush=True)
```

```python
# --- Coupled-Line Microwave Drives on Both Axes ---
drive_couplers, drive_branches, drive_trunks, drive_connections = [], [], [], []
drive_groups = [("H", row, bank) for row, bank in enumerate(horizontal_banks)]
drive_groups += [("V", col, bank) for col, bank in enumerate(vertical_banks)]
right_return_offset = max(
    float(qubits[row * columns + columns // 2].p.pos_x) - turn["x_mm"]
    for row, turn in enumerate(feedline_turns[1::2])
)
vertical_lane_offset = (
    max(right_return_offset, cpw_pitch + cpw_gap_mm) + cpw_pitch + cpw_gap_mm
)
drive_fillet = float(design.parse_value("15um"))
vertical_tee_offset = max(
    0.500, vertical_lane_offset + 2 * drive_fillet + cpw_slot_width
)
for axis, group_index, bank in drive_groups:
    row_tees, drive_pads = [], []
    labels = coupler_physical_table[
        (coupler_physical_table.axis == axis)
        & (coupler_physical_table["row" if axis == "H" else "column"] == group_index)
    ].coupler_class
    assert labels.nunique() == len(bank)
    for index, tc in enumerate(bank):
        cx, cy = float(tc.p.pos_x), float(tc.p.pos_y)
        if axis == "H":
            tx = cx - 0.430
            ty = horizontal_drive_y[group_index]
        else:
            tx, ty = cx - vertical_tee_offset, cy - 0.19925
        tee = CoupledLineTee(
            design,
            f"DRIVE_CLT_{axis}_{group_index}_{index}",
            options=Dict(
                pos_x=f"{tx}mm",
                pos_y=f"{ty}mm",
                orientation="180" if axis == "H" else "90",
                mirror=False,
                prime_width="8.5um",
                prime_gap="5.5um",
                second_width="8.5um",
                second_gap="5.5um",
                coupling_length="150um",
                coupling_space=f"{drive_tee_space}mm",
                down_length="30um",
                fillet="15um",
                open_termination=True,
            ),
        )
        row_tees.append(tee)
        drive_couplers.append(tee)
        p0, p1 = tee.pins.second_end.middle, tc.pins.drive.middle
        pin_inputs = Dict(
            start_pin=Dict(component=tee.name, pin="second_end"),
            end_pin=Dict(component=tc.name, pin="drive"),
        )
        if axis == "H":
            branch = RouteAnchors(
                design,
                f"CD_BRANCH_{axis}_{group_index}_{index}",
                options=Dict(
                    trace_width="8.5um",
                    trace_gap="5.5um",
                    fillet="15um",
                    pin_inputs=pin_inputs,
                    anchors=Dict({0: np.array([p0[0], p1[1]])}),
                    lead=Dict(start_straight="30um", end_straight="30um"),
                ),
            )
        else:
            branch = RouteStraight(
                design,
                f"CD_BRANCH_{axis}_{group_index}_{index}",
                options=Dict(
                    trace_width="8.5um", trace_gap="5.5um", pin_inputs=pin_inputs
                ),
            )
        drive_branches.append(branch)
    for side in (-1, 1):
        if axis == "H":
            px = side * (chip_x / 2 - 0.570)
            py = float(qubits[group_index * columns].p.pos_y) + 0.500
            angle = "0" if side == -1 else "180"
        else:
            px = float(qubits[group_index].p.pos_x) - 0.320
            py = side * (chip_y / 2 - 0.570)
            angle = "90" if side == -1 else "270"
        pad = LaunchpadWirebond(
            design,
            f"LP_COUPLER_{axis}_{group_index}_{side}",
            options=Dict(
                pos_x=f"{px}mm",
                pos_y=f"{py}mm",
                orientation=angle,
                trace_width="8.5um",
                trace_gap="5.5um",
                lead_length="25um",
                pad_width="170um",
                pad_height="230um",
                taper_height="240um",
                pad_gap="70um",
            ),
        )
        launchpads.append(pad)
        drive_pads.append(pad)
    entry, exit = (
        ("prime_end", "prime_start") if axis == "H" else ("prime_start", "prime_end")
    )
    joints = [(drive_pads[0], "tie", row_tees[0], entry)]
    joints += [(a, exit, b, entry) for a, b in zip(row_tees[:-1], row_tees[1:])]
    joints.append((row_tees[-1], exit, drive_pads[1], "tie"))
    for segment, (a, ap, b, bp) in enumerate(joints):
        p0, p1 = a.pins[ap].middle, b.pins[bp].middle
        if axis == "H":
            fanout = chip_x / 2 - 1.400
            points = (
                [p0, [-fanout, p0[1]], [-fanout, p1[1]], p1]
                if segment == 0
                else (
                    [p0, [fanout, p0[1]], [fanout, p1[1]], p1]
                    if segment == len(joints) - 1
                    else [p0, p1]
                )
            )
        else:
            lane_x = float(qubits[group_index].p.pos_x) - vertical_lane_offset
            if segment == 0:
                edge_y = p0[1] + 3 * drive_fillet
                lane_y = edge_y + abs(lane_x - p0[0]) / 2
                points = [
                    p0, [p0[0], edge_y], [lane_x, lane_y],
                    [lane_x, p1[1] - 0.200], [p1[0], p1[1] - 0.200], p1,
                ]
            elif segment == len(joints) - 1:
                return_y = (
                    float(qubits[(segment - 1) * columns + group_index].p.pos_y)
                    + readout_height + cpw_pitch + cpw_gap_mm
                )
                edge_y = p1[1] - 3 * drive_fillet
                lane_y = edge_y - abs(lane_x - p1[0]) / 2
                points = [
                    p0, [p0[0], return_y], [lane_x, return_y],
                    [lane_x, lane_y], [p1[0], edge_y], p1,
                ]
            else:
                return_y = (
                    float(qubits[(segment - 1) * columns + group_index].p.pos_y)
                    + readout_height + cpw_pitch + cpw_gap_mm
                )
                points = [
                    p0,
                    [p0[0], return_y],
                    [lane_x, return_y],
                    [lane_x, p1[1] - 0.200],
                    [p1[0], p1[1] - 0.200],
                    p1,
                ]
        drive_connections.append(
            (f"{axis}CDRIVE_{group_index}_{segment}", a, ap, b, bp, points)
        )

for axis in ("H", "V"):
    crossing_names = {
        component.id: cname
        for cname, component in design.components.items()
        if cname.startswith(("QCLINK_", "FEED_", "HCDRIVE_"))
    }
    paths = design.qgeometry.tables["path"]
    paths = paths[paths.component.isin(crossing_names) & ~paths["subtract"]]
    crossing_tree = STRtree(paths.geometry.to_list())
    for name, a, ap, b, bp, points in drive_connections:
        if name[0] != axis:
            continue
        nearby = crossing_tree.query(draw.LineString(points), predicate="intersects")
        crossing_prefixes = tuple(
            {crossing_names[cid] for cid in paths.iloc[nearby].component}
        )
        items = chip.split_at_crossings(
            design, points, name, over=crossing_prefixes, tag="ABD"
        )
        chain, pieces = [], []
        for kind, geometry in items:
            if kind == "bridge":
                chain.append((design.components[geometry], "a", "b"))
            else:
                piece = chip.PolylineCPW(
                    design,
                    name + (f"_{len(pieces)}" if pieces else ""),
                    options=Dict(
                        points=[list(map(float, p)) for p in geometry],
                        trace_width="8.5um",
                        trace_gap="5.5um",
                        fillet="15um",
                        min_segment="30um",
                    ),
                )
                pieces.append(piece)
                drive_trunks.append(piece)
                chain.append((piece, "start", "end"))
        for left, right in zip(chain[:-1], chain[1:]):
            design.connect_pins(left[0].id, left[2], right[0].id, right[1])
        design.connect_pins(a.id, ap, chain[0][0].id, chain[0][1])
        design.connect_pins(chain[-1][0].id, chain[-1][2], b.id, bp)
print(
    len(drive_couplers),
    "coupled-line drives;",
    len(launchpads),
    "perimeter signal launchers.",
    flush=True,
)
```

```python
# --- Multilayer Signal Crossovers ---
crossings = [c for c in design.components.values() if c.name.startswith("XO_")]
for old in list(design.components.values()):
    if not isinstance(old, Airbridge):
        continue
    name = "XO_" + old.name
    cx, cy = float(old.p.pos_x), float(old.p.pos_y)
    angle = float(old.p.orientation)
    length = float(old.p.crossover_length)
    feet_center = length / 2 + 0.0055
    peers = []
    for pin_name in ("a", "b"):
        net_id = old.pins[pin_name].net_id
        match = design.net_info[
            (design.net_info.net_id == net_id)
            & (design.net_info.component_id != old.id)
        ]
        for _, peer in match.iterrows():
            peers.append(
                (
                    pin_name,
                    design._components[int(peer.component_id)].name,
                    peer.pin_name,
                )
            )
    pin_points = {p: old.pins[p].points.copy() for p in ("a", "b")}
    design.delete_component(old.name)
    crossing = QComponent(
        design,
        name,
        options=Dict(
            pos_x=f"{cx}mm", pos_y=f"{cy}mm", orientation=str(angle), layer="1"
        ),
        make=False,
    )
    feet = draw.union(
        draw.rectangle(0.011, 0.011, -feet_center, 0),
        draw.rectangle(0.011, 0.011, feet_center, 0),
    )
    clearance = draw.union(
        draw.rectangle(0.022, 0.022, -feet_center, 0),
        draw.rectangle(0.022, 0.022, feet_center, 0),
    )
    span = draw.rectangle(2 * (feet_center + 0.005), 0.0085)
    scaffold = draw.rectangle(length + 0.001, 0.0195)
    contacts = draw.union(
        draw.rectangle(0.006, 0.005, -feet_center, 0),
        draw.rectangle(0.006, 0.005, feet_center, 0),
    )
    objects = draw.translate(
        draw.rotate([feet, clearance, span, scaffold, contacts], angle, origin=(0, 0)),
        cx,
        cy,
    )
    feet, clearance, span, scaffold, contacts = objects
    crossing.add_qgeometry("poly", {"base_landings": feet}, layer=1)
    crossing.add_qgeometry(
        "poly", {"landing_clearance": clearance}, layer=1, subtract=True
    )
    crossing.add_qgeometry("poly", {"reflow_scaffold": scaffold}, layer=7)
    crossing.add_qgeometry("poly", {"AlTiAl_overpass": span}, layer=8)
    crossing.add_qgeometry("poly", {"contacts": contacts}, layer=9)
    for p in ("a", "b"):
        middle = np.asarray(pin_points[p]).mean(axis=0)
        peer_name, peer_pin = next((n, pn) for pin, n, pn in peers if pin == p)
        route = design.components[peer_name]
        trace = route.qgeometry_table("path")
        if trace.empty:
            crossing.add_pin(p, np.asarray(pin_points[p]), width=0.011)
            continue
        coords = np.asarray(trace[~trace["subtract"]].iloc[0].geometry.coords)
        if np.linalg.norm(coords[-1] - middle) < np.linalg.norm(coords[0] - middle):
            coords = coords[::-1]
        normal = coords[1] - coords[0]
        normal /= np.linalg.norm(normal)
        crossing.add_pin(
            p,
            np.array([middle - 0.01 * normal, middle]),
            width=0.011,
            input_as_norm=True,
        )
    for p, peer_name, peer_pin in peers:
        design.connect_pins(crossing.id, p, design.components[peer_name].id, peer_pin)
    crossings.append(crossing)
print("PROCESS CROSSOVERS", len(crossings), flush=True)
```

```python
# --- Signal Ports and Protected Ground-Bond Landings ---
port_records = []
for pad in launchpads:
    x, y = float(pad.p.pos_x), float(pad.p.pos_y)
    horizontal = int(round(float(pad.p.orientation))) % 180 == 0
    edge = (
        ("left" if x < 0 else "right") if horizontal else ("bottom" if y < 0 else "top")
    )
    port_records.append(dict(component=pad.name, edge=edge, x_mm=x, y_mm=y))
port_table = pd.DataFrame(port_records)
ground_sheet = GroundContinuityRule(chip="main", layer=1)._ground_sheet(
    design, chip="main"
)
bond_lands, bond_records = {}, []
for edge, bank in port_table.groupby("edge"):
    along_y = edge in ("left", "right")
    positions = np.sort(bank.y_mm.to_numpy() if along_y else bank.x_mm.to_numpy())
    assert np.min(np.diff(positions)) >= port_pitch - 1e-9
    for gap, (low, high) in enumerate(zip(positions[:-1], positions[1:])):
        count = max(3, int(np.ceil((high - low - 0.400) / 0.200 - 1e-9)) + 1)
        for site, along in enumerate(np.linspace(low + 0.200, high - 0.200, count)):
            if along_y:
                x = (-1 if edge == "left" else 1) * (chip_x / 2 - 0.250)
                y = along
                landing = draw.rectangle(0.150, 0.075, x, y)
            else:
                x = along
                y = (-1 if edge == "bottom" else 1) * (chip_y / 2 - 0.250)
                landing = draw.rectangle(0.075, 0.150, x, y)
            assert ground_sheet.covers(landing.buffer(0.005)), (edge, gap, site)
            name = f"GND_{edge}_{gap}_{site}"
            bond_lands[name] = landing
            bond_records.append(
                dict(component=name, edge=edge, x_mm=x, y_mm=y, net="ground")
            )
ground_bond_component = QComponent(design, "GroundBondLands", make=False)
ground_bond_component.add_qgeometry("poly", bond_lands, layer=1)
ground_bond_table = pd.DataFrame(bond_records)
print(
    len(launchpads), "signal ports;", len(bond_lands), "ground-bond sites.", flush=True
)
```

```python
# --- Ground Straps Along the Filleted CPWs ---
if "GroundStraps" in design.components:
    design.delete_component("GroundStraps")

ground_sheet = GroundContinuityRule(chip="main", layer=1)._ground_sheet(
    design, chip="main"
)
ground_regions = [
    p for p in getattr(ground_sheet, "geoms", [ground_sheet]) if p.area > 1e-4
]
ground_graph = nx.Graph()
ground_graph.add_nodes_from(range(len(ground_regions)))
safe_regions = [region.buffer(-0.001) for region in ground_regions]
prepare(safe_regions)
region_tree = STRtree(safe_regions)
bridge_spans = [
    c.qgeometry_table("poly").query("subtract == False").geometry.union_all()
    for c in crossings
]
span_tree = STRtree(bridge_spans)
pending_spans = []
ground_straps = []
ground_masks = {layer: {} for layer in (1, 7, 8, 9)}
strap_pitch = float(design.parse_value("500um"))
for route in list(design.components.values()):
    if not isinstance(
        route, (chip.PolylineCPW, RouteStraight, RouteMeander, RouteAnchors)
    ):
        continue
    paths = route.qgeometry_table("path")
    points = np.asarray(paths[~paths["subtract"]].iloc[0].geometry.coords)
    steps = np.diff(points, axis=0)
    turns = steps[:-1, 0] * steps[1:, 1] - steps[:-1, 1] * steps[1:, 0]
    points = points[np.r_[True, np.abs(turns) > 1e-12, True]]
    radius = float(design.parse_value(route.options.get("fillet", "0um")))
    rounded = _fillet_centerline(points, radius)
    placements = _placements_along(rounded, strap_pitch, 0.100)
    for index, (cx, cy, angle) in enumerate(placements):
        feet_center = 0.0305
        feet = [draw.rectangle(0.011, 0.011, side * feet_center, 0) for side in (-1, 1)]
        span = draw.rectangle(0.071, 0.0085)
        scaffold = draw.rectangle(0.051, 0.0195)
        contacts = draw.union(
            *[draw.rectangle(0.006, 0.005, side * feet_center, 0) for side in (-1, 1)]
        )
        feet = draw.translate(draw.rotate(feet, angle, origin=(0, 0)), cx, cy)
        span, scaffold, contacts = draw.translate(
            draw.rotate([span, scaffold, contacts], angle, origin=(0, 0)), cx, cy
        )
        regions = []
        for foot in feet:
            matches = [
                int(i) for i in region_tree.query(foot) if safe_regions[i].covers(foot)
            ]
            if matches:
                regions.append(matches[0])
        if len(regions) != 2:
            continue
        envelope = draw.union(span, scaffold, *feet)
        if len(span_tree.query(envelope, predicate="dwithin", distance=0.003)):
            continue
        if any(envelope.distance(other) < 0.003 for other in pending_spans):
            continue
        name = f"GS_{route.name}_{index}"
        for layer, geometry in (
            (1, draw.union(*feet)),
            (7, scaffold),
            (8, span),
            (9, contacts),
        ):
            ground_masks[layer][name] = geometry
        ground_straps.append(name)
        ground_graph.add_edge(*regions)
        bridge_spans.append(envelope)
        pending_spans.append(envelope)
        if len(pending_spans) >= 32:
            span_tree = STRtree(bridge_spans)
            pending_spans.clear()
assert nx.is_connected(ground_graph), list(nx.connected_components(ground_graph))
ground_component = QComponent(design, "GroundStraps", make=False)
for layer, geometries in ground_masks.items():
    ground_component.add_qgeometry("poly", geometries, layer=layer)
print(
    f"{len(ground_straps)} ground straps connect all {len(ground_regions)} Nb ground regions."
)
```

```python
# --- Native Routing, Geometry, and Circuit Checks ---
assert len(qubits) == 200 and len(edge_specs) == 370
assert len(couplers) == len(resonators) == 200 and len(launchpads) == 100
assert len(design.qgeometry.tables["junction"]) == 570
assert len(transmon_couplers) == 370 and len(coupler_links) == 740
assert len(drive_couplers) == len(drive_branches) == 370
assert not any(name.startswith(("FLUX_", "SQUID_")) for name in design.components)
assert nx.is_connected(ground_graph)
for table_name in ("poly", "path", "junction"):
    table = design.qgeometry.tables[table_name]
    assert table.geometry.is_valid.all(), table_name
    assert not table.geometry.is_empty.any(), table_name
    xmin, ymin, xmax, ymax = table.geometry.total_bounds
    assert xmin >= -chip_x / 2 and xmax <= chip_x / 2, table_name
    assert ymin >= -chip_y / 2 and ymax <= chip_y / 2, table_name
for pad in launchpads:
    metal = pad.qgeometry_table("poly").query("subtract == False").geometry.union_all()
    xmin, ymin, xmax, ymax = metal.bounds
    assert np.isclose(
        min(xmin + chip_x / 2, chip_x / 2 - xmax, ymin + chip_y / 2, chip_y / 2 - ymax),
        0.100,
        atol=1e-6,
    ), pad.name
for q in qubits:
    table = q.qgeometry_table("poly")
    base = table[(table.layer == 1) & ~table["subtract"]].geometry.union_all()
    bandages = table[table.layer == 5].geometry.union_all()
    lower = table[table.layer == 3].geometry.union_all()
    upper = table[table.layer == 4].geometry.union_all()
    overlaps = table[table.layer == 6].geometry.union_all()
    island_contact = table[
        table.name.isin(("circle_inner", "island_contact"))
    ].geometry.union_all()
    ground_contact = table[table.name == "ground_contact"].geometry.union_all()
    assert ground_contact.intersection(ground_sheet).area > 1e-5, q.name
    assert island_contact.distance(ground_contact) > 0.010, q.name
    for pad in table[table.name.str.startswith("contact_")].geometry:
        assert island_contact.distance(pad) >= 0.005, q.name
    assert np.isclose(
        lower.intersection(upper).area, overlaps.area, atol=1e-15, rtol=1e-4
    ), q.name
    for bandage in bandages.geoms:
        assert base.covers(bandage) or ground_sheet.covers(bandage), q.name
        assert lower.union(upper).buffer(1e-9).covers(bandage), q.name
    wire = q.qgeometry_table("path").query("name == 'high_Lk_shunt'").iloc[0]
    rounded = draw.LineString(
        _fillet_centerline(np.asarray(wire.geometry.coords), wire.fillet)
    )
    assert rounded.is_simple, q.name
    contact_neighborhoods = draw.union(
        *[draw.Point(*p).buffer(0.006) for p in (rounded.coords[0], rounded.coords[-1])]
    )
    unexpected_contact = (
        rounded.buffer(wire_width / 2)
        .intersection(lower.union(upper))
        .difference(contact_neighborhoods)
    )
    assert unexpected_contact.area < 1e-12, q.name
    for endpoint in (rounded.coords[0], rounded.coords[-1]):
        assert bandages.covers(draw.Point(*endpoint)), q.name
for tc in transmon_couplers:
    table = tc.qgeometry_table("poly")
    base = table[(table.layer == 1) & ~table["subtract"]].geometry.union_all()
    lower = table[table.layer == 3].geometry.union_all()
    upper = table[table.layer == 4].geometry.union_all()
    bandages = table[table.layer == 5].geometry.union_all()
    overlap = table[table.layer == 6].geometry.union_all()
    assert np.isclose(
        lower.intersection(upper).area, overlap.area, rtol=1e-4, atol=1e-15
    ), tc.name
    assert base.buffer(1e-9).covers(bandages), tc.name
    assert lower.union(upper).buffer(1e-9).covers(bandages), tc.name
    top = table.query("name == 'pad_top'").geometry.iloc[0]
    bottom = table.query("name == 'pad_bot'").geometry.iloc[0]
    assert np.isclose(top.distance(bottom), 0.020, atol=1e-9), tc.name
    assert all(
        tc.pins[name].net_id != 0 for name in ("left", "right", "drive")
    ), tc.name
for route in coupler_links + drive_branches:
    lengths = np.linalg.norm(np.diff(route.get_points(), axis=0), axis=1)
    radius = float(route.p.fillet)
    assert min(lengths[0], lengths[-1]) >= radius - 1e-9, route.name
    if len(lengths) > 2:
        assert min(lengths[1:-1]) >= 2 * radius - 1e-9, route.name
# Check electrical separation between the shared trunks and their coupled branches.
drive_metal = []
for component in drive_trunks + drive_couplers:
    paths = component.qgeometry_table("path")
    paths = paths[~paths["subtract"]]
    if isinstance(component, CoupledLineTee):
        paths = paths[paths.name == "prime_cpw"]
    for _, path in paths.iterrows():
        centerline = _fillet_centerline(np.asarray(path.geometry.coords), path.fillet)
        drive_metal.append(
            draw.LineString(centerline).buffer(path.width / 2, cap_style=2)
        )
drive_metal += [
    c.qgeometry_table("poly")
    .query("layer == 1 and subtract == False")
    .geometry.union_all()
    for c in crossings
    if c.name.startswith("XO_ABD_")
]
drive_tree = STRtree(drive_metal)
for tee, branch in zip(drive_couplers, drive_branches):
    paths = tee.qgeometry_table("path")
    primary = paths[paths.name == "prime_cpw"].iloc[0]
    secondary = paths[paths.name == "second_cpw"].iloc[0]
    primary_metal = primary.geometry.buffer(primary.width / 2, cap_style=2)
    secondary_metal = draw.LineString(
        _fillet_centerline(np.asarray(secondary.geometry.coords), secondary.fillet)
    ).buffer(secondary.width / 2, cap_style=2)
    branch_metal = draw.LineString(
        _fillet_centerline(branch.get_points(), float(branch.p.fillet))
    ).buffer(float(branch.p.trace_width) / 2, cap_style=2)
    clearance = float(tee.p.prime_gap + tee.p.second_gap + tee.p.coupling_space)
    assert np.isclose(
        primary_metal.distance(secondary_metal), clearance, atol=1e-9
    ), tee.name
    assert not len(
        drive_tree.query(secondary_metal.union(branch_metal), predicate="intersects")
    ), tee.name
    assert tee.p.open_termination is True, tee.name
    assert all(
        tee.pins[name].net_id != 0
        for name in ("prime_start", "prime_end", "second_end")
    ), tee.name
# --- Complete Transmon-Assisted Nearest-Neighbor Graph ---
interaction_graph = nx.Graph()
interaction_graph.add_nodes_from(q.name for q in qubits)
for record in coupler_physical_records:
    interaction_graph.add_edge(
        record["left_qubit"], record["right_qubit"], axis=record["axis"]
    )
assert interaction_graph.number_of_nodes() == 200
assert interaction_graph.number_of_edges() == 370 and nx.is_connected(interaction_graph)
assert coupler_physical_table.axis.value_counts().to_dict() == {"H": 190, "V": 180}
assert not any(name.startswith("BUS_") for name in design.components)

# --- Parallel CPW Clearance ---
slot_paths = design.qgeometry.tables["path"].query("subtract and layer == 1")
straight_runs, run_widths, run_paths, run_owners = [], [], [], []
for path_id, path in enumerate(slot_paths.itertuples()):
    points = np.asarray(path.geometry.simplify(1e-10).coords)
    steps = np.diff(points, axis=0)
    lengths = np.linalg.norm(steps, axis=1)
    directions = steps / lengths[:, None]
    turn_angles = np.arccos(
        np.clip(np.sum(directions[1:] * directions[:-1], axis=1), -1, 1)
    )
    trim = np.minimum(
        path.fillet * np.tan(turn_angles / 2),
        np.minimum(lengths[:-1], lengths[1:]) / 2,
    )
    starts = points[:-1] + np.r_[0, trim][:, None] * directions
    stops = points[1:] - np.r_[trim, 0][:, None] * directions
    for start, stop in zip(starts, stops):
        if np.linalg.norm(stop - start) > 1e-6:
            straight_runs.append([start, stop])
            run_widths.append(path.width)
            run_paths.append(path_id)
            run_owners.append(path.component)

runs = np.asarray(straight_runs)
paths = linestrings(runs)
widths = np.asarray(run_widths)
owners = np.asarray(run_owners)
path_ids = np.asarray(run_paths)
tee_ids = [tee.id for tee in couplers + drive_couplers]
left, right = STRtree(paths).query(paths, predicate="dwithin", distance=2 * cpw_pitch)
same_tee = (owners[left] == owners[right]) & np.isin(owners[left], tee_ids)
keep = (left < right) & (path_ids[left] != path_ids[right]) & ~same_tee
left, right = left[keep], right[keep]
steps = runs[:, 1] - runs[:, 0]
lengths = np.linalg.norm(steps, axis=1)
directions = steps / lengths[:, None]
parallel = (
    abs(np.sum(directions[left] * directions[right], axis=1)) >= np.cos(np.deg2rad(3))
)
left, right = left[parallel], right[parallel]
projection = np.einsum("ijk,ik->ij", runs[right] - runs[left, :1], directions[left])
overlap = np.minimum(lengths[left], projection.max(axis=1)) - np.maximum(
    0, projection.min(axis=1)
)
left, right = left[overlap > 1e-6], right[overlap > 1e-6]
spacing = distance(paths[left], paths[right]) - (widths[left] + widths[right]) / 2
if len(spacing):
    closest = int(np.argmin(spacing))
    pair = [design._components[int(owners[i])].name for i in (left[closest], right[closest])]
    print(
        f"Minimum unintended parallel CPW gap: {1000 * spacing[closest]:.1f} µm "
        f"({pair[0]} and {pair[1]}).",
        flush=True,
    )
    assert spacing[closest] >= cpw_clearance - 1e-6, (pair, spacing[closest])

# --- Readout-to-Coupling CPW Clearance ---
routing_envelopes, routing_names = [], []
for component in (
    coupler_links + drive_trunks + drive_branches + drive_couplers
    + transmon_couplers
):
    for _, path in (
        component.qgeometry_table("path").query("subtract == True").iterrows()
    ):
        points = _fillet_centerline(np.asarray(path.geometry.coords), path.fillet)
        routing_envelopes.append(
            draw.LineString(points).buffer(path.width / 2, cap_style=2)
        )
        routing_names.append(component.name)
    if isinstance(component, (TransmonPocket, TransmonPocket6)):
        pocket = component.qgeometry_table("poly").query("name == 'rect_pk'")
        routing_envelopes.append(pocket.geometry.iloc[0])
        routing_names.append(component.name + " pocket")
for component in crossings:
    if component.name.startswith("XO_ABD_"):
        routing_envelopes.append(component.qgeometry_table("poly").geometry.union_all())
        routing_names.append(component.name)
routing_tree = STRtree(routing_envelopes)
clearance_records = []
for q, route, tee in zip(qubits, resonators, couplers):
    path = route.qgeometry_table("path").query("subtract == True").iloc[0]
    envelope = draw.LineString(
        _fillet_centerline(route.get_points(), path.fillet)
    ).buffer(path.width / 2, cap_style=2)
    secondary = tee.qgeometry_table("path").query("name == 'second_cpw_sub'").iloc[0]
    secondary_envelope = draw.LineString(
        _fillet_centerline(np.asarray(secondary.geometry.coords), secondary.fillet)
    ).buffer(secondary.width / 2, cap_style=2)
    escape = draw.Point(float(q.p.pos_x), float(q.p.pos_y)).buffer(0.320, quad_segs=64)
    envelope = envelope.union(secondary_envelope)
    local = envelope.intersection(escape)
    local_gap_um = 1000 * local.distance(
        routing_envelopes[int(routing_tree.nearest(local))]
    )
    assert local_gap_um >= 50 - 1e-3, (route.name, local_gap_um)
    corridor = envelope.difference(escape)
    nearest = int(routing_tree.nearest(corridor))
    clearance = corridor.distance(routing_envelopes[nearest])
    assert clearance >= cpw_clearance - 1e-6, (
        route.name,
        clearance,
    )
    clearance_records.append(dict(
        qubit=q.name,
        corridor_gap_um=1000 * clearance,
        nearest_coupling_feature=routing_names[nearest],
        local_fanout_gap_um=local_gap_um,
    ))
clearance_table = pd.DataFrame(clearance_records)
print(
    f"Minimum readout corridor gap: {clearance_table.corridor_gap_um.min():.1f} um. "
    f"Minimum local fanout gap: {clearance_table.local_fanout_gap_um.min():.1f} um.",
    flush=True,
)

for component in crossings + [ground_component]:
    table = component.qgeometry_table("poly")
    lower = table[(table.layer == 1) & ~table["subtract"]].geometry.union_all()
    upper = table[table.layer == 8].geometry.union_all()
    contacts = table[table.layer == 9].geometry.union_all()
    assert lower.buffer(1e-9).covers(contacts), component.name
    assert upper.buffer(1e-9).covers(contacts), component.name
for route in resonators:
    distances = np.linalg.norm(np.diff(route.get_points(), axis=0), axis=1)
    radius = float(route.p.fillet)
    assert np.min(distances[[0, -1]]) >= radius - 1e-9, route.name
    assert np.min(distances[1:-1]) >= 2 * radius - 1e-9, route.name
length_table = pd.DataFrame(length_records)
length_table["model_GHz"] = phase_velocity / (4 * length_table.length_mm * 1e-3) / 1e9
assert np.max(np.abs(length_table.model_GHz - length_table.target_GHz)) < 1e-5
physical_table = pd.DataFrame(physical_records)
for row in range(rows):
    for half in range(2):
        bank = physical_table[
            (physical_table.row == row) & (physical_table.column // 10 == half)
        ]
        assert len(bank.frequency_class.unique()) == 10
        assert np.min(np.diff(np.sort(bank.model_f01_GHz))) > 0.050

waivers = [
    Waiver(
        rule="ground-continuity",
        reason=f"The verified elevated-strap graph connects all {len(ground_regions)} base-ground regions; the native rule sees layer 1 only.",
    )
]
drc = validate(design, rules=[*DEFAULT_RULES, *SHAPE_RULES], waivers=waivers)
print(drc.report())
drc.raise_if_errors()
summary = pd.DataFrame(
    [
        ("Star fluxonium sites", 200),
        ("Horizontal / vertical transmon couplers", "190 / 180"),
        ("Physical single junctions", 570),
        ("Flux-bias lines", 0),
        ("Coupled readout resonators", len(resonators)),
        ("Shared readout/control feedlines", 20),
        ("Coupler charge-drive through lines", 30),
        ("Coupled-line charge-drive couplers", len(drive_couplers)),
        ("Perimeter signal launchers", len(launchpads)),
        ("Signal crossings", len(crossings)),
        ("Ground straps", len(ground_straps)),
        ("Ground-bond landing sites", len(bond_lands)),
        ("Chip X / Y (mm)", f"{chip_x:g} / {chip_y:g}"),
    ],
    columns=["Layout item", "Count or size"],
)
display(summary)
display(clearance_table.round(3))
display(port_table.groupby("edge").size().rename("Signal ports").to_frame())
display(length_table.round(6))
```

```python
# --- Native Cheesing and GDSII Export ---
out = Path.cwd() / OUTPUT_FILENAME
gds = QGDSRenderer(design, initiate=False)
gds.options.precision = "1e-9"
gds.options.tolerance = "1e-6"
gds.options.max_points = "8190"
gds.options.fabricate = "True"
gds.options.short_segments_to_not_fillet = "False"
gds.options.path_filename = None
gds.options.cheese = QGDSRenderer.default_options.cheese.copy()
gds.options.no_cheese = QGDSRenderer.default_options.no_cheese.copy()
gds.options.cheese.edge_nocheese = "112.5um"
gds.options.no_cheese.buffer = CHEESE_KEEPOUT
gds.options.cheese.view_in_file = Dict(
    main={layer: layer == 1 for layer in range(1, 10)}
)
gds.options.no_cheese.view_in_file = Dict(main={layer: False for layer in range(1, 10)})

# Physical junction electrodes already occupy layers 3 and 4.
junction_table = design.qgeometry.tables["junction"].copy(deep=True)
design.qgeometry.tables["junction"] = junction_table.iloc[:0].copy()
previous_log_level = qm.logger.level
qm.logger.setLevel(logging.ERROR)
try:
    with warnings.catch_warnings():
        warnings.filterwarnings(
            "ignore",
            category=RuntimeWarning,
            message=r"^Error in boolean operation\.$",
            module=r"^qiskit_metal\.renderers\.renderer_gds\.gds_boolean$",
        )
        warnings.filterwarnings(
            "ignore",
            category=FutureWarning,
            message="The behavior of DataFrame concatenation.*",
        )
        export_status = gds.export_to_gds(str(out))
finally:
    design.qgeometry.tables["junction"] = junction_table
    qm.logger.setLevel(previous_log_level)
assert export_status and out.is_file() and out.stat().st_size > 0

# --- GDS Readback and Physical Mask Checks ---
verified_gds = gdstk.read_gds(str(out))
top_cells = verified_gds.top_level()
assert len(top_cells) == 1
exported_polygons = top_cells[0].get_polygons()
assert {polygon.layer for polygon in exported_polygons} == set(range(1, 10))
assert np.isclose(verified_gds.unit, 1e-3, rtol=0, atol=1e-12)
assert np.isclose(verified_gds.precision, 1e-9, rtol=0, atol=1e-15)
assert sum(p.layer == 2 for p in exported_polygons) == 200
assert sum(p.layer == 6 for p in exported_polygons) == 570
expected_wire_area = physical_table.drawn_wire_length_um.sum() / 1000 * wire_width
actual_wire_area = sum(p.area() for p in exported_polygons if p.layer == 2)
assert np.isclose(actual_wire_area, expected_wire_area, rtol=0.01, atol=0)
actual_overlaps = gdstk.boolean(
    [p for p in exported_polygons if p.layer == 3],
    [p for p in exported_polygons if p.layer == 4],
    "and",
    precision=1e-6,
)
assert len(actual_overlaps) == 570
assert np.allclose(
    np.sort([p.area() * 1e6 for p in actual_overlaps]),
    np.sort(
        np.r_[
            physical_table.nominal_junction_area_um2,
            coupler_physical_table.nominal_junction_area_um2,
        ]
    ),
    rtol=0.02,
    atol=0,
)
contacts = [p for p in exported_polygons if p.layer == 9]
for layer in (1, 8):
    missing = gdstk.boolean(
        contacts,
        [p for p in exported_polygons if p.layer == layer],
        "not",
        precision=1e-6,
    )
    assert sum(p.area() for p in missing) < 1e-10, layer
bandages = [p for p in exported_polygons if p.layer == 5]
for layers in ((1,), (3, 4)):
    missing = gdstk.boolean(
        bandages,
        [p for p in exported_polygons if p.layer in layers],
        "not",
        precision=1e-6,
    )
    assert sum(p.area() for p in missing) < 1e-10, layers
bond_polygons = [
    gdstk.Polygon(np.asarray(g.exterior.coords)) for g in bond_lands.values()
]
missing = gdstk.boolean(
    bond_polygons, [p for p in exported_polygons if p.layer == 1], "not", precision=1e-6
)
assert sum(p.area() for p in missing) < 1e-10

# --- Signal and Ground Bond Map ---
signal_bond_records = []
for pad in launchpads:
    angle = np.radians(float(pad.p.orientation))
    center = np.array([float(pad.p.pos_x), float(pad.p.pos_y)])
    center -= (float(pad.p.taper_height) + float(pad.p.pad_height) / 2) * np.array(
        [np.cos(angle), np.sin(angle)]
    )
    signal_bond_records.append(
        dict(
            component=pad.name,
            edge=(
                ("left" if center[0] < 0 else "right")
                if abs(np.cos(angle)) > 0.5
                else ("bottom" if center[1] < 0 else "top")
            ),
            x_mm=center[0],
            y_mm=center[1],
            net="signal",
        )
    )
bond_map = pd.concat(
    [pd.DataFrame(signal_bond_records), ground_bond_table], ignore_index=True
)
bond_map = bond_map.sort_values(["edge", "x_mm", "y_mm"])
bond_map_path = out.with_name(out.stem + "_Bond_Map.csv")
bond_map.to_csv(bond_map_path, index=False)
print(
    f"GDSII: {out.name} ({out.stat().st_size:,} bytes); 200 fluxonium sites, 370 fixed transmon couplers; layers 1–9."
)
print(
    "Readback: 200 continuous superinductors, 570 junction overlaps, complete bridge contacts and bond lands."
)
display(FileLink(str(out)))
display(FileLink(str(bond_map_path)))
```
